# Missão Aurora Singer

## Relatório Operacional de Pré-Decolagem

Projeto desenvolvido para a Fase 1 da FIAP.

O objetivo é analisar dados simulados de telemetria, verificar as condições da nave, realizar uma análise energética e apoiar a interpretação dos resultados com Inteligência Artificial.

> As faixas e valores utilizados neste projeto são premissas acadêmicas definidas pelo grupo para fins de simulação.

```python
# Dados simulados da Missão Aurora Singer

# Integridade estrutural
c1 = 1
c2 = 1
c3 = 1
c4 = 1
c5 = 1

# Níveis de energia
bateria = 90
sistema_critico = 100

# Pressão dos tanques
p_lox = 45
p_lh2 = 40
p_helio = 3500
p_camara = 1500

# Status dos módulos críticos
m1 = 1
m2 = 1
m3 = 1
m4 = 1
m5 = 1
m6 = 1

# Temperaturas
t_camara = 3000
t_lox = 100
t_lh2 = 25
```

```python
# Verificações da telemetria

# Integridade estrutural
ok_estrutural = (
    c1 == 1 and
    c2 == 1 and
    c3 == 1 and
    c4 == 1 and
    c5 == 1
)

# Níveis de energia
ok_energia = (
    bateria >= 70 and
    sistema_critico >= 70
)

# Pressão dos tanques
ok_lox = 30 <= p_lox <= 50
ok_lh2 = 30 <= p_lh2 <= 50
ok_helio = 2175 <= p_helio <= 4350
ok_camara_p = 150 <= p_camara <= 3000

ok_pressao = (
    ok_lox and
    ok_lh2 and
    ok_helio and
    ok_camara_p
)

# Status dos módulos críticos
ok_modulos = (
    m1 == 1 and
    m2 == 1 and
    m3 == 1 and
    m4 == 1 and
    m5 == 1 and
    m6 == 1
)

# Temperaturas
ok_temp_camara = 2700 <= t_camara <= 3600
ok_temp_lox = 90 <= t_lox <= 165
ok_temp_lh2 = 20 <= t_lh2 <= 33

ok_temperaturas = (
    ok_temp_camara and
    ok_temp_lox and
    ok_temp_lh2
)
```

```python
print("=== RELATÓRIO DE VERIFICAÇÃO - AURORA SINGER ===")

print("Integridade estrutural:", "APROVADO" if ok_estrutural else "REPROVADO")
print("Níveis de energia:", "APROVADO" if ok_energia else "REPROVADO")
print("Pressão dos tanques:", "APROVADO" if ok_pressao else "REPROVADO")
print("Status dos módulos críticos:", "APROVADO" if ok_modulos else "REPROVADO")
print("Temperaturas:", "APROVADO" if ok_temperaturas else "REPROVADO")

if (
    ok_estrutural
    and ok_energia
    and ok_pressao
    and ok_modulos
    and ok_temperaturas
):
    print("\nRESULTADO FINAL: GO PARA LANÇAMENTO")
else:
    print("\nRESULTADO FINAL: NOT-GO - LANÇAMENTO ABORTADO")
```

```python
# Análise energética

capacidade_total = 5000      # kWh
consumo_decolagem = 1500    # kWh
perdas = 0.05               # 5%

# A carga atual utiliza o valor da bateria definido na telemetria
carga_atual = bateria / 100

energia_disponivel = capacidade_total * carga_atual
energia_perdida = energia_disponivel * perdas
energia_util = energia_disponivel - energia_perdida
energia_restante = energia_util - consumo_decolagem

print("=== ANÁLISE ENERGÉTICA ===")
print("Capacidade total do sistema:", capacidade_total, "kWh")
print("Carga atual da bateria:", bateria, "%")
print("Energia disponível:", energia_disponivel, "kWh")
print("Perdas energéticas:", perdas * 100, "%")
print("Energia perdida:", energia_perdida, "kWh")
print("Energia útil:", energia_util, "kWh")
print("Consumo estimado na decolagem:", consumo_decolagem, "kWh")
print("Energia restante após a decolagem:", energia_restante, "kWh")
```
